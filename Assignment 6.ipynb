Problem 1: Smart Parking Lot Management System

```python
def manage_parking(capacity, logs):
    # Calculate current count based on IN/OUT logs
    current_parked = 0
    for action in logs:
        if action == "IN":
            current_parked += 1
        elif action == "OUT":
            current_parked -= 1
    
    # Check capacity status
    status = "Full" if current_parked >= capacity else "Available"
    
    print(f"Currently Parked Vehicles: {current_parked}")
    print(f"Parking Status: {status}")

# Sample Input
manage_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

Problem 2: Online Food Delivery Time Estimator

```python
def estimate_delivery(distance, traffic, weather):
    # Base time: 5 mins per km
    base_time = distance * 5
    
    # Traffic Adjustments
    traffic_delays = {"Low": 0, "Medium": 10, "High": 20}
    
    # Weather Adjustments
    weather_delays = {"Clear": 0, "Rainy": 15, "Stormy": 30}
    
    total_eta = base_time + traffic_delays.get(traffic, 0) + weather_delays.get(weather, 0)
    
    print(f"Estimated Delivery Time: {total_eta} minutes")

# Sample Input
estimate_delivery(8, "High", "Rainy")
```

Problem 3: Movie Theatre Seat Occupancy Analyzer

```python
def analyze_occupancy(total_seats, booked_seats_list):
    booked_count = len(booked_seats_list)
    occupancy_pct = (booked_count / total_seats) * 100
    
    status = "Housefull" if occupancy_pct == 100 else "Almost Full" if occupancy_pct >= 70 else "Available"
    
    print(f"Occupancy: {occupancy_pct}%")
    print(f"Show Status: {status}")

# Sample Input (Simulating 150 entries)
analyze_occupancy(200, [1] * 150)
```

Problem 4: Cloud Server Load Classification System

```python
def classify_server_load(readings):
    avg_load = sum(readings) / len(readings)
    
    if avg_load < 50:
        status = "Normal"
    elif 50 <= avg_load <= 80:
        status = "Warning"
    else:
        status = "Critical"
        
    print(f"Average CPU Load: {avg_load}%")
    print(f"Server Status: {status}")

# Sample Input
classify_server_load([45, 60, 70, 85, 90])
```

Problem 5: Smart Classroom Resource Usage Monitor

```python
def monitor_resources(usage_dict):
    # Threshold for overuse: 8 hours
    threshold = 8
    overused = [res for res, hours in usage_dict.items() if hours > threshold]
    
    alert = "Yes" if overused else "No"
    
    print(f"Overused Resources: {', '.join(overused)}")
    print(f"Energy Alert: {alert}")

# Sample Input
monitor_resources({"Projector": 6, "AC": 9, "Lights": 4})
```

Problem 6: Online Event Registration Capacity Controller

```python
def manage_registrations(capacity, total_registrations):
    if total_registrations > capacity:
        confirmed = capacity
        waitlisted = total_registrations - capacity
        status = "Closed"
    else:
        confirmed = total_registrations
        waitlisted = 0
        status = "Open"
        
    print(f"Confirmed Registrations: {confirmed}")
    print(f"Waitlisted Users: {waitlisted}")
    print(f"Registration Status: {status}")

# Sample Input
manage_registrations(100, 105)
```
